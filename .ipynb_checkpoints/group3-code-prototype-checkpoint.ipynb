 1. journal names produced manually: 
     1.1 WOS journal names - provided in the 2018 WOS csv file
     1.2. new data source journal names - user must provide a data source with a column containing journal names
 2. journal names produced automatically:
     1.1 WOS journal names - provided in the 2018 WOS csv file
     1.2 new data source journal names - user must provide a data source with a column containing journal names

Use cases:
- WOS file provided
- reports provided
- authors provided

```python
from annoy import AnnoyIndex
import random

f = 40
t = AnnoyIndex(f, 'angular')  # Length of item vector that will be indexed
print(t)
```

```python
for i in range(1000):
    v = [random.gauss(0, 1) for z in range(f)]
    t.add_item(i, v)
print(t)
print(v)
```

```python
t.build(10) # 10 trees
t.save('test.ann')

# ...

u = AnnoyIndex(f, 'angular')
u.load('test.ann') # super fast, will just mmap the file
print(u.get_nns_by_item(0, 1000)) # will find the 1000 nearest neighbors
```
